## Step 1 - Build a robust grouping function

The function below reads rows by header names, normalizes primary color and product type,
counts rows (including duplicates), and returns sorted summary dictionaries.


```python
def summarize_palettes(csv_path):
    groups = {}
    with open(csv_path, 'r', encoding='utf-8', newline='') as f:
        reader = csv.DictReader(f)
        if not reader.fieldnames:
            return []
        for row in reader:
            primary = (row.get('Primary (Hex)') or '').strip().upper()
            product_type = (row.get('Product Type') or '').strip()

            if primary not in groups:
                groups[primary] = {
                    'primary': primary,
                    'count': 0,
                    'product_types': set(),
                }

            groups[primary]['count'] += 1
            groups[primary]['product_types'].add(product_type)

    summary = []
    for item in groups.values():
        summary.append({
            'primary': item['primary'],
            'count': item['count'],
            'product_types': sorted(item['product_types']),
        })

    summary.sort(key=lambda x: (-x['count'], x['primary']))
    return summary
```

```python
palette_summary = summarize_palettes('/inputs/colors.csv')
palette_summary
```
