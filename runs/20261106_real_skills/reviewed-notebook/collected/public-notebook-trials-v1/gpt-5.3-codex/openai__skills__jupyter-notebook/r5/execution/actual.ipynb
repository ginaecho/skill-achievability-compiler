## Learning goals

## Outline

2. Define `summarize_palettes(csv_path)`.
3. Run the function on the provided palette catalog.


```python
import csv
```

## Implementation
The function below reads rows with `csv.DictReader`, strips surrounding whitespace from
`Primary (Hex)` and `Product Type`, uppercases primary colors, groups rows, and returns
sorted summary objects.


```python
def summarize_palettes(csv_path):
    groups = {}
    with open(csv_path, "r", encoding="utf-8", newline="") as handle:
        reader = csv.DictReader(handle)
        if not reader.fieldnames:
            return []
        for row in reader:
            primary = (row.get("Primary (Hex)", "") or "").strip().upper()
            product_type = (row.get("Product Type", "") or "").strip()
            if primary not in groups:
                groups[primary] = {
                    "primary": primary,
                    "count": 0,
                    "product_types": set(),
                }
            groups[primary]["count"] += 1
            groups[primary]["product_types"].add(product_type)
    summary = []
    for item in groups.values():
        summary.append({
            "primary": item["primary"],
            "count": item["count"],
            "product_types": sorted(item["product_types"]),
        })
    summary.sort(key=lambda item: (-item["count"], item["primary"]))
    return summary
```

```python
palette_summary = summarize_palettes("/inputs/colors.csv")
```
