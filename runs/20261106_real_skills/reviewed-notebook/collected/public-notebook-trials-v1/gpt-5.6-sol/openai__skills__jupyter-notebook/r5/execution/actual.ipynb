```python
from __future__ import annotations

import csv
from pathlib import Path
```

## Step 1: Group and normalize the rows

`csv.DictReader` addresses columns by header, so column order does not matter. For each row, the function strips both requested fields, uppercases the primary color, counts every row (including duplicates), and keeps unique product types in a set. The return step converts those sets to sorted lists and orders groups by descending count, then ascending primary text.


```python
def summarize_palettes(csv_path):
    groups = {}

    with open(csv_path, mode="r", encoding="utf-8", newline="") as csv_file:
        for row in csv.DictReader(csv_file):
            primary = row["Primary (Hex)"].strip().upper()
            product_type = row["Product Type"].strip()
            group = groups.setdefault(primary, {"count": 0, "product_types": set()})
            group["count"] += 1
            group["product_types"].add(product_type)

    ordered_groups = sorted(
        groups.items(),
        key=lambda item: (-item[1]["count"], item[0]),
    )
    return [
        {
            "primary": primary,
            "count": details["count"],
            "product_types": sorted(details["product_types"]),
        }
        for primary, details in ordered_groups
    ]
```
