## Step 1 â€” Prepare the standard-library tools

`csv.DictReader` maps each row to the CSV headers, so column order does not matter. A dictionary of sets will collect product types without duplicates.


```python
from pathlib import Path
import csv

CATALOG_PATH = Path('/inputs/colors.csv')
```

## Step 2 â€” Normalize, group, and sort

For each row, surrounding whitespace is removed from both relevant fields. Primary colors are uppercased before grouping. The final sort uses descending counts and then ordinary lexicographic order for ties.


```python
def summarize_palettes(csv_path):
    """Group catalog rows by normalized primary color."""
    groups = {}

    with open(csv_path, mode='r', encoding='utf-8', newline='') as catalog_file:
        rows = csv.DictReader(catalog_file)
        for row in rows:
            primary = row['Primary (Hex)'].strip().upper()
            product_type = row['Product Type'].strip()

            if primary not in groups:
                groups[primary] = {'count': 0, 'product_types': set()}
            groups[primary]['count'] += 1
            groups[primary]['product_types'].add(product_type)

    summaries = [
        {
            'primary': primary,
            'count': details['count'],
            'product_types': sorted(details['product_types']),
        }
        for primary, details in groups.items()
    ]
    return sorted(summaries, key=lambda item: (-item['count'], item['primary']))
```

## Step 3 â€” Summarize the catalog

The next cell calls the reusable function on the public catalog. Each result has only `primary`, `count`, and `product_types`.


```python
palette_summary = summarize_palettes('/inputs/colors.csv')
palette_summary
```
