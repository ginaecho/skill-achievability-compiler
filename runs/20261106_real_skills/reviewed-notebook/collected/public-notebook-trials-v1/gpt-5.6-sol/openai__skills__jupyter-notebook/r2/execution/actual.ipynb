# Tutorial: Palette catalog summaries

## Audience

## Prerequisites

- A UTF-8 CSV file with `Primary (Hex)` and `Product Type` headers

## Learning goals
By the end, you can read rows by header, normalize selected text fields, group duplicate records, and produce deterministically sorted summaries.


## Outline

2. Apply it to the public palette catalog.
3. Inspect compact, derived facts about the result.
4. Extend the approach in the exercises.


```python
import csv
from collections import defaultdict


def summarize_palettes(csv_path):
    """Return normalized palette groups from a UTF-8 CSV catalog."""
    groups = defaultdict(lambda: {"count": 0, "product_types": set()})

    with open(csv_path, mode="r", encoding="utf-8", newline="") as catalog:
        for row in csv.DictReader(catalog):
            primary = row["Primary (Hex)"].strip().upper()
            product_type = row["Product Type"].strip()
            groups[primary]["count"] += 1
            groups[primary]["product_types"].add(product_type)

    summary = [
        {
            "primary": primary,
            "count": values["count"],
            "product_types": sorted(values["product_types"]),
        }
        for primary, values in groups.items()
    ]
    return sorted(summary, key=lambda item: (-item["count"], item["primary"]))
```

## Group and sort the catalog

`csv.DictReader` maps each row using the CSV headers, including quoted fields that contain commas. The function strips both grouping fields, uppercases the primary color, keeps every row in the count, and uses a set to deduplicate product types. The final sort uses a negative count for descending frequency and the primary text for ascending tie-breaking.


```python
palette_summary = summarize_palettes('/inputs/colors.csv')
palette_summary[:5]
```
