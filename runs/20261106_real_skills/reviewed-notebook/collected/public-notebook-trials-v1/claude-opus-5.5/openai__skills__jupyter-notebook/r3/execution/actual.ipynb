```python
# Setup cell: standard library only
from __future__ import annotations

import csv
from collections import defaultdict

CATALOG_PATH = '/inputs/colors.csv'
CATALOG_PATH
```

## Step 1 - Group rows by primary color

The function reads the CSV by its headers, strips whitespace from `Primary (Hex)` and `Product Type`, uppercases the primary color, and groups rows by that normalized value. Each group reports its row count (duplicates included) and its sorted unique product types. Groups are ordered by count descending, then primary ascending. A header-only file gives an empty list.


```python
def summarize_palettes(csv_path):
    counts = defaultdict(int)
    product_types = defaultdict(set)
    with open(csv_path, newline='', encoding='utf-8') as handle:
        for row in csv.DictReader(handle):
            primary = (row['Primary (Hex)'] or '').strip().upper()
            product = (row['Product Type'] or '').strip()
            counts[primary] += 1
            product_types[primary].add(product)
    summary = [
        {'primary': primary, 'count': count, 'product_types': sorted(product_types[primary])}
        for primary, count in counts.items()
    ]
    summary.sort(key=lambda group: (-group['count'], group['primary']))
    return summary
```

## Step 2 - Summarize the catalog

Apply the function to the real catalog and look at the first few groups instead of printing everything.


```python
palette_summary = summarize_palettes(CATALOG_PATH)
[(group['primary'], group['count']) for group in palette_summary[:5]]
```
