```python
# Setup: standard library only
import csv
from collections import defaultdict

CATALOG_PATH = '/inputs/colors.csv'
```

## Step 1 - Group rows by normalized primary color

Each row's `Primary (Hex)` is stripped and uppercased; `Product Type` is stripped.
Every row counts (duplicates included); product types are collected as a sorted unique list.
Groups are ordered by count descending, then primary ascending.


```python
def summarize_palettes(csv_path):
    counts = defaultdict(int)
    types = defaultdict(set)
    with open(csv_path, newline='', encoding='utf-8') as handle:
        for row in csv.DictReader(handle):
            primary = (row.get('Primary (Hex)') or '').strip().upper()
            product_type = (row.get('Product Type') or '').strip()
            counts[primary] += 1
            types[primary].add(product_type)
    summary = [
        {'primary': primary, 'count': counts[primary], 'product_types': sorted(types[primary])}
        for primary in counts
    ]
    summary.sort(key=lambda item: (-item['count'], item['primary']))
    return summary
```

## Step 2 - Summarize the catalog

Apply the function to the public catalog and look at the largest groups.


```python
palette_summary = summarize_palettes('/inputs/colors.csv')
print(len(palette_summary), 'groups')
for group in palette_summary[:3]:
    print(group['primary'], group['count'], len(group['product_types']))
```
